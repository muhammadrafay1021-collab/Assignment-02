```python
inventory = {
    "The Great Gatsby": {"price": 10.99, "quantity": 5}, 
    "Rich dad poor dad": {"price": 8.99, "quantity": 8}, 
    "Harry Potter": {"price": 12.50, "quantity": 2}
}
inventory["Atomic Habits"] = {"price": 14.99, "quantity": 0}
book_titles = list(inventory.keys())
print("All Book Titles:", book_titles)
for book, details in list(inventory.items()):
    if details["quantity"] == 0:
        inventory.pop(book)

print("\nUpdated Inventory after removing out-of-stock book:")
print(inventory)
```

```python


char = input("Enter a single letter: ")


if len(char) == 1 and char.isalpha():
    
    vowels = "aeiouAEIOU"
    
    if char in vowels:
        print(f"{char} is a Vowel")
    else:
        print(f"{char} is a Consonant")
else:
    print("Please enter valid alphabet")
```
